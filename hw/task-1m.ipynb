**Признаки**

* `age` — возраст в годах;
* `sex` — пол (1 = мужчина; 0 = женщина);
* `cp` — тип боли в груди (0: typical angina, 1: atypical angina, 2: non-anginal pain, 3: asymptomatic);
* `trestbps` — артериальное давление в состоянии покоя (в мм рт. ст.);
* `chol` — уровень холестерина в крови (в мг/дл);
* `fbs` — уровень сахара в крови натощак > 120 мг/дл (1 = true; 0 = false);

* `thalach` — максимальная достигнутая частота сердечных сокращений;
* `exang` — стенокардия, вызванная физической нагрузкой (1 = да; 0 = нет);
* `oldpeak` — депрессия ST, вызванная физической нагрузкой относительно покоя;
* `slope` — наклон пика упражнения сегмента ST (0: восходящий, 1: плоский, 2: нисходящий);
* `ca` — количество основных сосудов (0-3), окрашенных флуороскопией;
* `thal` — талассемия (3 = норма; 6 = исправленный дефект; 7 = обратимый дефект);
* `target` — наличие сердечного заболевания (0 = нет; 1-4: различные стадии заболевания).
